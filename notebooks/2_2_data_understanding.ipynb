# Этап 2.2. Подбор и изучение источников данных

**Тема:** исследование банковских и биржевых данных для обнаружения аномалий и паттернов финансовых транзакций.


## Краткий каталог

1. **MoneyData (XLSX, реляционные банковские операции).** Категориальные: `Transaction Type`, `Category`, `Location City`, `Location Country`. Числовые: `Debit Amount`, `Credit Amount`, `Balance`. Время: `Transaction Date`. Лицензия CC BY 4.0.
2. **SEC EDGAR XBRL instance (XML, отчётность JPMorgan).** Категориальные: эмитент/CIK, форма, таксономия, концепт, единица/контекст. Числовые факты отчётности: `Assets`, `Liabilities`, `NetIncomeLoss`; для одного отчёта они выбираются по периоду, контексту и единице. Время: период факта и дата подачи.
3. **Binance Spot API `/api/v3/klines` (JSON, одна свеча пары к USDT).** Категориальные: биржа, тип рынка, пара, базовый/котируемый актив, интервал запроса. Числовые поля ответа: open, high, low, close, volume, quote volume, число сделок. Время: open/close time.
4. **Financial News Dataset (текст статей с JSON-метаданными).** Категориальные: домен, язык, тикеры, отрасли, издатель. Числовые поля источника: negative/neutral/positive sentiment и цены тикеров до/в день/после публикации. Время: `date_publish`. Лицензия CC BY-NC-SA 4.0.

Внешние источники и фактические периоды покрытия будут загружаться и проверяться в пункте 2.3. Подробное обоснование, ограничения и ссылки находятся в `docs/2_2_sources.md`.

```python
sources = [
    {"название": "MoneyData: банковская операция", "тип": "реляционный XLSX", "категориальные": 4, "числовые": 3, "время": True},
    {"название": "SEC XBRL: отчёт эмитента за период", "тип": "XML", "категориальные": 5, "числовые": 3, "время": True},
    {"название": "Binance Spot: свеча торговой пары", "тип": "API JSON", "категориальные": 6, "числовые": 7, "время": True},
    {"название": "Финансовая новостная статья", "тип": "неструктурированный текст", "категориальные": 5, "числовые": 3, "время": True},
]

types_required = {"реляционный XLSX", "XML", "API JSON", "неструктурированный текст"}
assert len(sources) >= 4
assert all(item["категориальные"] >= 3 for item in sources)
assert all(item["числовые"] >= 3 for item in sources)
assert all(item["время"] for item in sources)
assert types_required == {item["тип"] for item in sources}

print(f"Требуемые типы источников представлены: {len(sources)} из 4")
print("Для каждого каталожного объекта указано категориальных характеристик: не менее 3")
print("Для каждого каталожного объекта указано числовых характеристик: не менее 3")
```

## Ограничения выбранного состава

Банковский файл не содержит отдельной колонки валюты. Binance BTC/USDT — криптовалютный спотовый рынок, а не рынок акций; это ограничение явно учитывается при формулировке выводов. Отчётность SEC описывает публичного эмитента, а не анонимизированного владельца счёта. Совпадение месяцев позволяет сравнить временной контекст, но не доказывает причинность.

Наборы с Kaggle допустимы, если проверены лицензия и происхождение конкретного набора. Kaggle JSON-файл не заменяет API с JSON-ответом. В текущий набор добавлен уже имеющийся лицензированный Mendeley-файл, поэтому дублирующий банковский Kaggle-датасет не нужен.
